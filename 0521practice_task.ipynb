```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv("E:\Job\The Living Stones\Ag Trends in South America (2013-2023).csv")
```

```python
data
```

```python
data_aggregate = data.groupby(['Year', 'Crop Type']).agg({
	'Production Volume (in tons)': 'sum',
	'Actual Sales Volume': 'sum'
}).reset_index()

data_aggregate.head()
```

```python
x = data_aggregate['Crop Type']
y = data_aggregate['Production Volume (in tons)']

plt.figure(figsize=(10, 8))
plt.bar(x, y, label='Value',color='lightblue')
plt.xlabel('Crop Type')
plt.ylabel('Production Volume (in tons)')
plt.title('Production Volume by Crop Type')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
